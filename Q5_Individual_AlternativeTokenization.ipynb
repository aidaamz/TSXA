This notebook implements **spaCy** (with a custom tokenizer rule) as the alternative tokenisation approach, and compares it against the group's Q1 approach (NLTK `word_tokenize`):

**Data source:** `Data_1.txt` (the same file used in the group's Q1 notebook, for a fair comparison)


```python
print("=" * 60)
print("CELL 1: SETUP & LOAD DATA")
print("=" * 60)

# spaCy is the alternative library for this task; NLTK is imported only to
# recreate the group's Q1 baseline (94 tokens) for a fair, live comparison
try:
    import spacy
    import pandas as pd
    import nltk
    from nltk.tokenize import word_tokenize
    print("✓ All libraries imported successfully")
except ImportError as e:
    print(f"✗ Import error: {e}")

nltk.download('punkt', quiet=True)
nltk.download('punkt_tab', quiet=True)

# Load spaCy's small English pipeline (industry-standard NLP library)
nlp = spacy.load("en_core_web_sm")

# Read the same text used in the group's Q1 notebook
try:
    with open('Data_1.txt', 'r') as f:
        text = f.read()
except FileNotFoundError:
    print("✗ Error: 'Data_1.txt' not found. Make sure it is in the same folder as this notebook.")
    text = ""

# Recreate the group's NLTK tokenisation as the baseline for comparison
tokens_nltk = word_tokenize(text)

print(f"\nCharacter count           : {len(text)}")
print(f"NLTK word_tokenize count  : {len(tokens_nltk)}  (baseline from Q1)")
print("✓ Setup complete")
print("=" * 60)
```

```python
print("=" * 60)
print("STEP 1: TOKENISATION USING spaCy (DEFAULT PIPELINE)")
print("=" * 60)

# Run the text through spaCy's pipeline; doc is a sequence of Token objects
doc_default = nlp(text)
tokens_spacy_default = [token.text for token in doc_default]

# Print every token with its index number (repr shows whitespace tokens clearly)
for i, token in enumerate(tokens_spacy_default, start=1):
    print(f"{i:5}. {token!r}")

print(f"\n  -> Total tokens: {len(tokens_spacy_default)}")
print("=" * 60)
```

```python
print("=" * 60)
print("STEP 2: IDENTIFYING A PROBLEM -- THE 'open-class' CASE")
print("=" * 60)

# Locate the tokens spaCy produced for the compound term "open-class"
idx = tokens_spacy_default.index("open")
print("spaCy (default) splits 'open-class' into:")
print("  ", tokens_spacy_default[idx:idx + 3])

print("\nWhy this is a problem:")
print("  'open-class' is a single linguistic concept (a type of classification")
print("  task). Breaking it into 'open', '-', 'class' loses that meaning --")
print("  'class' is now indistinguishable from every other generic occurrence")
print("  of 'class' in the text (e.g. 'class label').")

print("\nFor comparison:")
print(f"  spaCy (default) total tokens : {len(tokens_spacy_default)}")
print(f"  NLTK word_tokenize total      : {len(tokens_nltk)}  (keeps 'open-class' intact)")
print("=" * 60)
```

```python
print("=" * 60)
print("STEP 3: FIXING THE PROBLEM WITH A CUSTOM TOKENIZER RULE")
print("=" * 60)

from spacy.util import compile_infix_regex


HYPHEN_GROUP = "(?:-|–|—|--|---|——|~)"
custom_infixes = [
    pattern for pattern in nlp.Defaults.infixes
    if HYPHEN_GROUP not in pattern
]

nlp.tokenizer.infix_finditer = compile_infix_regex(custom_infixes).finditer

# Re-tokenise the same text with the modified tokenizer
doc_custom = nlp(text)
tokens_spacy_custom = [token.text for token in doc_custom]

# Print every token with its index number
for i, token in enumerate(tokens_spacy_custom, start=1):
    print(f"{i:5}. {token!r}")

print(f"\n  -> Total tokens: {len(tokens_spacy_custom)}")
print("  -> 'open-class' is now ONE token:", "open-class" in tokens_spacy_custom)

# Sanity checks: confirm we did NOT break other useful hyphen behaviour
sanity = nlp("a value of 10-20 and a well-known, state-of-the-art fact.")
print("\nSanity check on other hyphen patterns:")
print("  ", [t.text for t in sanity])
print("=" * 60)
```

```python
print("=" * 60)
print("STEP 4: COMPARISON WITH THE GROUP'S Q1 APPROACH (NLTK)")
print("=" * 60)

comparison_df = pd.DataFrame({
    "Method": ["NLTK word_tokenize (Q1, group)", "spaCy (default)", "spaCy (custom infix rule)"],
    "Token Count": [len(tokens_nltk), len(tokens_spacy_default), len(tokens_spacy_custom)],
    "'open-class' handling": [
        "1 token (preserved)",
        "3 tokens: open / - / class",
        "1 token (preserved)"
    ],
    "Newline token": ["Discarded", "Kept as a token", "Kept as a token"]
})

print(comparison_df.to_string(index=False))

print("\nThe single extra token in spaCy (custom) vs NLTK is the newline character:")
print("  '\\n' in spaCy (custom) tokens :", '\n' in tokens_spacy_custom)
print("  '\\n' in NLTK tokens           :", '\n' in tokens_nltk)
print("=" * 60)
```

## Q5 — Compare, Contrast & Justify

### Approach Chosen: spaCy, with a Custom Tokenizer Rule

[spaCy](https://spacy.io/) is an industrial-strength NLP library widely used in production systems (search, chatbots, information extraction). Unlike NLTK's `word_tokenize()`, spaCy's tokeniser is **rule-based and fully configurable** — its splitting behaviour ("infix", "prefix" and "suffix" rules) can be inspected and modified directly in code.

- **Step 1 (default spaCy):** Tokenising `Data_1.txt` with spaCy's default English pipeline produces **97 tokens** — *more* than NLTK's 94, and *worse* on one specific point: the compound term `open-class` is broken into three tokens, `open`, `-`, `class`, destroying its meaning as a single technical term.
- **Step 2 (custom rule):** spaCy's default infix rules include a pattern that splits any `ALPHA-HYPHEN-ALPHA` sequence. By removing **only that one pattern** and recompiling the tokenizer, hyphenated compounds such as `open-class`, `well-known`, and `state-of-the-art` are kept as single tokens — while numeric ranges such as `10-20` still split correctly, because that is governed by a *separate*, untouched rule. This produces **95 tokens**.

### Compare and Contrast with the Group's NLTK Approach (2 marks)

| Aspect | NLTK `word_tokenize` (Q1, group) | spaCy (default) | spaCy (custom rule) |
|---|---|---|---|
| Token count | 94 | 97 | 95 |
| `open-class` | 1 token | 3 tokens (`open`, `-`, `class`) | 1 token |
| `\n` (line break in source text) | discarded | kept as a token | kept as a token |
| Rule customisation | Fixed, "black box" (hard-coded in C) | Fully configurable via `infix`/`prefix`/`suffix` regex lists | Fully configurable |
| Underlying approach | Penn Treebank conventions | Rule-based, regex-driven | Rule-based, regex-driven (modified) |

The two approaches agree on almost everything — punctuation separation, contractions, sentence-final stops. The **only substantive difference**, once the custom rule is applied, is that spaCy retains the newline character `\n` from the source file as its own token, while NLTK silently discards all whitespace.

### Is spaCy (Custom) Better, Worse, or Just Different? (5 marks)

**Better:**
The default spaCy tokenizer actually performed *worse* than NLTK on `open-class` (3 tokens vs 1) — but because spaCy exposes its tokenisation rules as editable regex lists, this weakness could be **fixed directly in code**, which is not possible with NLTK's `word_tokenize()` (its rules are hard-coded in C). This is a real advantage for **domain-specific text analytics**: if a corpus regularly contains hyphenated technical terms (e.g. "open-class", "part-of-speech", "co-occurrence"), the tokenizer can be *tuned* to preserve them, directly improving the quality of any downstream bag-of-words or TF-IDF model.

**Worse / a trade-off:**
spaCy's default pipeline (`en_core_web_sm`) is a ~12 MB statistical model — far heavier than NLTK's lightweight `word_tokenize()` — which is unnecessary overhead for a simple tokenisation task on a short text. Also, retaining `\n` as a token means spaCy's output contains a "noise" token that NLTK does not produce; this would need to be filtered out in a later stop-word/punctuation removal step (Q1 Part 3), adding an extra cleaning step that NLTK's pipeline does not require.

**Just Different:**
Both approaches are *correct*, in the sense that they follow internally consistent, linguistically motivated rules. The difference is one of philosophy: NLTK's `word_tokenize()` is a fixed, "best-practice default" suitable for general English text, whereas spaCy treats tokenisation as a **configurable first stage of an NLP pipeline**, designed to be adapted to the vocabulary of a specific domain or corpus.

### Conclusion

For this short, general-English text, NLTK's `word_tokenize()` and a *correctly configured* spaCy tokenizer produce almost identical results (94 vs 95 tokens, differing only by the retained newline character). The key finding is that spaCy's **default** configuration is actually less suitable than NLTK for this corpus — but spaCy's strength is that this weakness is fixable through rule customisation, making it the more suitable choice for **domain-specific text analytics**, where preserving compound technical terms matters.
